```python
import json
import requests

QUERY_PATH = "queries/pdb_query.json"
SEARCH_URL = "https://search.rcsb.org/rcsbsearch/v2/query"

with open(QUERY_PATH, "r") as f:
    query = json.load(f)

# Ensure request_options/paginate exists
query.setdefault("request_options", {})
query["request_options"].setdefault("paginate", {})

rows = 1000  # 1000 is a safe, common page size
start = 0
all_ids = []

while True:
    query["request_options"]["paginate"]["start"] = start
    query["request_options"]["paginate"]["rows"] = rows

    resp = requests.post(SEARCH_URL, json=query)
    if not resp.ok:
        print("Status:", resp.status_code)
        print(resp.text)
        resp.raise_for_status()

    data = resp.json()
    result_set = data.get("result_set", [])
    batch_ids = [r["identifier"] for r in result_set]
    all_ids.extend(batch_ids)

    print(f"Fetched {len(batch_ids)} (total {len(all_ids)})")

    # If we got fewer than rows, we're done
    if len(batch_ids) < rows:
        break

    start += rows

# De-dupe, preserve order
seen = set()
pdb_ids = []
for pid in all_ids:
    if pid not in seen:
        seen.add(pid)
        pdb_ids.append(pid)

with open("data/pdb_ids.txt", "w") as out:
    out.write(','.join(pdb_ids))

print(f"Retrieved {len(pdb_ids)} unique PDB IDs")

```
